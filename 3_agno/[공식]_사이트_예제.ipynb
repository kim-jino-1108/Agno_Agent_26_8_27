# Lab 1: Basic Agent




```python
"""
기본 에이전트
=============================
기본 에이전트 빠른 시작 예제.
"""
# 에이전트 클래스를 가져온다.
from agno.agent import Agent

# Responses API(/v1/responses)를 사용하는 OpenAI 모델 클래스를 가져온다.
from agno.models.openai import OpenAIResponses

# ---------------------------------------------------------------------------
# 에이전트 생성
# ---------------------------------------------------------------------------
agent = Agent(
    # 에이전트를 식별하기 위한 이름
    name="Quickstart Agent",
    # 사용할 모델을 id로 지정해 생성한다.
    model=OpenAIResponses(id="gpt-5-mini"),
)

# ---------------------------------------------------------------------------
# 에이전트 실행
# ---------------------------------------------------------------------------
agent.print_response(
        "인사하고, 한 문장으로 자기소개를 해줘.", stream=True  # 에이전트에 전달할 입력 메시지 (한글)
    )
```

# Lab 2: Agent With Instructions

https://docs.agno.com/examples/agents/basics/agent-with-instructions


```python
"""
지시문이 있는 에이전트
=============================
지시문이 있는 에이전트 빠른 시작 예제.
"""
# 에이전트 클래스를 가져온다.
from agno.agent import Agent

# Responses API(/v1/responses)를 사용하는 OpenAI 모델 클래스를 가져온다.
from agno.models.openai import OpenAIResponses

# ---------------------------------------------------------------------------
# 에이전트 지시문
# ---------------------------------------------------------------------------
# 에이전트의 응답 방식을 정의하는 시스템 지시문 (한글)
# 문자열 끝의 백슬래시는 줄바꿈을 없애 불필요한 앞뒤 공백 줄이 생기지 않게 한다.
instructions = """\
당신은 간결한 어시스턴트입니다.
가능한 경우 정확히 3개의 불릿 포인트로 답하십시오.\
"""

# ---------------------------------------------------------------------------
# 에이전트 생성
# ---------------------------------------------------------------------------
agent = Agent(
    # 에이전트를 식별하기 위한 이름
    name="Instruction-Tuned Agent",
    # 사용할 모델을 id로 지정해 생성한다.
    model=OpenAIResponses(id="gpt-5-mini"),
    # 위에서 정의한 지시문을 시스템 프롬프트로 전달한다.
    instructions=instructions,
)

# ---------------------------------------------------------------------------
# 에이전트 실행
# ---------------------------------------------------------------------------
# 입력 메시지를 전달하고 응답을 콘솔에 출력한다. stream=True는 생성되는 대로 이어서 출력한다.
agent.print_response(
        "파이썬 디버깅 작업 흐름을 어떻게 개선할 수 있을까?", stream=True  # 에이전트에 전달할 입력 메시지 (한글)
 )
```

# Lab3  Agent With Tools


https://docs.agno.com/examples/agents/basics/agent-with-tools



```python
"""
도구를 사용하는 에이전트
=============================
도구를 사용하는 에이전트 빠른 시작 예제.
"""
# 에이전트 클래스를 가져온다.
from agno.agent import Agent

# Responses API(/v1/responses)를 사용하는 OpenAI 모델 클래스를 가져온다.
from agno.models.openai import OpenAIResponses

# DuckDuckGo 웹 검색 기능을 제공하는 도구 모음을 가져온다.
from agno.tools.duckduckgo import DuckDuckGoTools

# ---------------------------------------------------------------------------
# 에이전트 생성
# ---------------------------------------------------------------------------
agent = Agent(
    # 에이전트를 식별하기 위한 이름
    name="Tool-Enabled Agent",
    # 사용할 모델을 id로 지정해 생성한다.
    model=OpenAIResponses(id="gpt-5-mini"),
    # 에이전트가 호출할 수 있는 도구 목록: DuckDuckGo 웹 검색
    tools=[DuckDuckGoTools()],
)

# ---------------------------------------------------------------------------
# 에이전트 실행
# ---------------------------------------------------------------------------
# 입력 메시지를 전달하고 응답을 콘솔에 출력한다. stream=True는 생성되는 대로 이어서 출력한다.
agent.print_response(
        "최근 AI safety headline  하나를 찾아서 요약해줘.", stream=True  # 에이전트에 전달할 입력 메시지 (한글)
    )
```
